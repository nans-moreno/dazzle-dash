```python
import pandas as pd
import numpy as np

# 1. Chargement des données
airlines = pd.read_csv('airlines.csv')
airports = pd.read_csv('airports.csv')
flights = pd.read_csv('flights.csv')
```

```python
# 2. Inspection rapide
print("Aperçu airlines.csv :")
display(airlines.head())
print("Aperçu airports.csv :")
display(airports.head())
print("Aperçu flights.csv :")
display(flights.head())
```

```python
# 3. Vérification des valeurs manquantes
print("Valeurs manquantes airlines.csv :\n", airlines.isnull().sum())
print("Valeurs manquantes airports.csv :\n", airports.isnull().sum())
print("Valeurs manquantes flights.csv :\n", flights.isnull().sum())
```

```python
# 4. Nettoyage airlines.csv
# Suppression des doublons
airlines = airlines.drop_duplicates()
# Suppression des lignes où le code ou le nom est manquant
airlines = airlines.dropna(subset=['IATA_CODE', 'AIRLINE'])
```

```python
# 5. Nettoyage airports.csv
airports = airports.drop_duplicates()
# Suppression des lignes où le code ou le nom d'aéroport est manquant
airports = airports.dropna(subset=['IATA_CODE', 'AIRPORT'])
# Conversion des coordonnées en float
airports['LATITUDE'] = pd.to_numeric(airports['LATITUDE'], errors='coerce')
airports['LONGITUDE'] = pd.to_numeric(airports['LONGITUDE'], errors='coerce')
```

```python
# 6. Nettoyage flights.csv
flights = flights.drop_duplicates()
# Suppression des vols sans code compagnie ou aéroport
flights = flights.dropna(subset=['AIRLINE', 'ORIGIN_AIRPORT', 'DESTINATION_AIRPORT'])
# Conversion des colonnes numériques
num_cols = ['YEAR','MONTH','DAY','DAY_OF_WEEK','SCHEDULED_DEPARTURE','DEPARTURE_TIME','DEPARTURE_DELAY',
            'TAXI_OUT','WHEELS_OFF','SCHEDULED_TIME','ELAPSED_TIME','AIR_TIME','DISTANCE','WHEELS_ON',
            'TAXI_IN','SCHEDULED_ARRIVAL','ARRIVAL_TIME','ARRIVAL_DELAY','DIVERTED','CANCELLED',
            'AIR_SYSTEM_DELAY','SECURITY_DELAY','AIRLINE_DELAY','LATE_AIRCRAFT_DELAY','WEATHER_DELAY']
for col in num_cols:
    if col in flights.columns:
        flights[col] = pd.to_numeric(flights[col], errors='coerce')

# Remplacement des chaînes vides par NaN
flights = flights.replace('', np.nan)
```

```python
# 7. Vérification post-nettoyage
print("airlines.csv nettoyé :\n", airlines.info())
print("airports.csv nettoyé :\n", airports.info())
print("flights.csv nettoyé :\n", flights.info())
```

```python
# 8. Sauvegarde des fichiers nettoyés
airlines.to_csv('airlines_clean.csv', index=False)
airports.to_csv('airports_clean.csv', index=False)
flights.to_csv('flights_clean.csv', index=False)

print("Nettoyage terminé et fichiers sauvegardés.")
```
