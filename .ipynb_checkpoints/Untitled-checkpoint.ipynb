```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import warnings
from datetime import datetime

warnings.filterwarnings('ignore')

# Configuration de l'affichage
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.float_format', '{:.2f}'.format)
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette("Set2")

print("="*80)
print("ANALYSE EXPLORATOIRE DES DONNÉES DE VOLS")
print("="*80)
print(f"Démarrage de l'analyse : {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
print("="*80 + "\n")
```

```python
# 1. CHARGEMENT DES DONNÉES
# ============================================================================

def load_data():
    """Charge les trois fichiers CSV"""
    print("📂 Chargement des données...")
    
    try:
        airlines = pd.read_csv('airlines.csv')
        airports = pd.read_csv('airports.csv')
        flights = pd.read_csv('flights.csv', 
                            dtype={'ORIGIN_AIRPORT': str, 'DESTINATION_AIRPORT': str},
                            low_memory=False)
        
        print(f"✅ Données chargées avec succès:")
        print(f"   - Vols: {len(flights):,} lignes")
        print(f"   - Compagnies: {len(airlines)} lignes")
        print(f"   - Aéroports: {len(airports)} lignes\n")
        
        return flights, airlines, airports
    
    except Exception as e:
        print(f"❌ Erreur lors du chargement: {e}")
        return None, None, None
```

```python
# 2. EXPLORATION INITIALE
# ============================================================================

def explore_structure(df, name):
    """Affiche la structure d'un DataFrame"""
    print(f"\n{'='*80}")
    print(f"STRUCTURE - {name}")
    print("="*80)
    print(f"Dimensions: {df.shape[0]:,} lignes × {df.shape[1]} colonnes\n")
    print("Types de données:")
    print(df.dtypes)
    print("\nAperçu des premières lignes:")
    print(df.head())
    print("\nStatistiques descriptives:")
    print(df.describe())

def analyze_missing(df, name):
    """Analyse des valeurs manquantes"""
    print(f"\n{'='*80}")
    print(f"VALEURS MANQUANTES - {name}")
    print("="*80)
    
    missing = pd.DataFrame({
        'Colonne': df.columns,
        'Manquantes': df.isnull().sum(),
        'Pourcentage': (df.isnull().sum() / len(df) * 100).round(2)
    }).sort_values('Manquantes', ascending=False)
    
    missing_filtered = missing[missing['Manquantes'] > 0]
    
    if len(missing_filtered) > 0:
        print(missing_filtered.to_string())
        
        # Graphique des valeurs manquantes
        if len(missing_filtered) > 0:
            plt.figure(figsize=(12, 6))
            plt.barh(missing_filtered['Colonne'].head(15), 
                    missing_filtered['Pourcentage'].head(15))
            plt.xlabel('Pourcentage de valeurs manquantes (%)')
            plt.title(f'Valeurs manquantes - {name}')
            plt.tight_layout()
            plt.savefig(f'missing_values_{name}.png', dpi=300, bbox_inches='tight')
            print(f"   📊 Graphique sauvegardé: missing_values_{name}.png")
            plt.close()
    else:
        print("✅ Aucune valeur manquante")
```

```python
# 3. ENRICHISSEMENT DES DONNÉES
# ============================================================================

def enrich_data(flights, airlines, airports):
    """Enrichit le DataFrame flights avec les noms et coordonnées"""
    print("\n📊 Enrichissement des données...")
    
    # Création des dictionnaires de mapping
    airline_dict = dict(zip(airlines['IATA_CODE'], airlines['AIRLINE']))
    airport_dict = dict(zip(airports['IATA_CODE'], airports['AIRPORT']))
    city_dict = dict(zip(airports['IATA_CODE'], airports['CITY']))
    state_dict = dict(zip(airports['IATA_CODE'], airports['STATE']))
    lat_dict = dict(zip(airports['IATA_CODE'], airports['LATITUDE']))
    lon_dict = dict(zip(airports['IATA_CODE'], airports['LONGITUDE']))
    
    # Ajout des noms complets
    flights['AIRLINE_NAME'] = flights['AIRLINE'].map(airline_dict)
    flights['ORIGIN_NAME'] = flights['ORIGIN_AIRPORT'].map(airport_dict)
    flights['ORIGIN_CITY'] = flights['ORIGIN_AIRPORT'].map(city_dict)
    flights['ORIGIN_STATE'] = flights['ORIGIN_AIRPORT'].map(state_dict)
    flights['DESTINATION_NAME'] = flights['DESTINATION_AIRPORT'].map(airport_dict)
    flights['DESTINATION_CITY'] = flights['DESTINATION_AIRPORT'].map(city_dict)
    flights['DESTINATION_STATE'] = flights['DESTINATION_AIRPORT'].map(state_dict)
    
    # Ajout des coordonnées
    flights['ORIGIN_LAT'] = flights['ORIGIN_AIRPORT'].map(lat_dict)
    flights['ORIGIN_LON'] = flights['ORIGIN_AIRPORT'].map(lon_dict)
    flights['DEST_LAT'] = flights['DESTINATION_AIRPORT'].map(lat_dict)
    flights['DEST_LON'] = flights['DESTINATION_AIRPORT'].map(lon_dict)
    
    # Conversion des colonnes numériques
    numeric_cols = ['DEPARTURE_DELAY', 'ARRIVAL_DELAY', 'MONTH', 'DAY', 'DAY_OF_WEEK']
    for col in numeric_cols:
        if col in flights.columns:
            flights[col] = pd.to_numeric(flights[col], errors='coerce')
    
    # Ajout de colonnes dérivées
    mois_noms = {1: 'Janvier', 2: 'Février', 3: 'Mars', 4: 'Avril', 5: 'Mai', 6: 'Juin',
                 7: 'Juillet', 8: 'Août', 9: 'Septembre', 10: 'Octobre', 11: 'Novembre', 12: 'Décembre'}
    
    saisons = {1: 'Hiver', 2: 'Hiver', 3: 'Hiver', 4: 'Printemps', 5: 'Printemps', 6: 'Printemps',
               7: 'Été', 8: 'Été', 9: 'Été', 10: 'Automne', 11: 'Automne', 12: 'Automne'}
    
    flights['MONTH_NAME'] = flights['MONTH'].map(mois_noms)
    flights['SEASON'] = flights['MONTH'].map(saisons)
    flights['ON_TIME'] = (flights['ARRIVAL_DELAY'] <= 15).astype(int)
    flights['DELAYED'] = (flights['ARRIVAL_DELAY'] > 15).astype(int)
    flights['CANCELLED'] = flights.get('CANCELLED', 0).fillna(0).astype(int)
    
    print("✅ Données enrichies avec succès\n")
    return flights
```

```python
# 4. STATISTIQUES GLOBALES
# ============================================================================

def global_statistics(flights):
    """Calcule et affiche les statistiques globales"""
    print("\n" + "="*80)
    print("STATISTIQUES GLOBALES")
    print("="*80)
    
    total_vols = len(flights)
    cancelled = flights['CANCELLED'].sum()
    delayed = flights['DELAYED'].sum()
    ontime = flights['ON_TIME'].sum()
    
    avg_delay_arr = flights['ARRIVAL_DELAY'].mean()
    median_delay_arr = flights['ARRIVAL_DELAY'].median()
    avg_delay_dep = flights['DEPARTURE_DELAY'].mean()
    
    print(f"\n🛫 VOLUME:")
    print(f"   Total de vols: {total_vols:,}")
    print(f"   Compagnies: {flights['AIRLINE'].nunique()}")
    print(f"   Aéroports origine: {flights['ORIGIN_AIRPORT'].nunique()}")
    print(f"   Aéroports destination: {flights['DESTINATION_AIRPORT'].nunique()}")
    
    print(f"\n⏰ PONCTUALITÉ:")
    print(f"   Vols à l'heure (≤15 min): {ontime:,} ({ontime/total_vols*100:.2f}%)")
    print(f"   Vols en retard (>15 min): {delayed:,} ({delayed/total_vols*100:.2f}%)")
    print(f"   Vols annulés: {cancelled:,} ({cancelled/total_vols*100:.2f}%)")
    
    print(f"\n⏱️ DÉLAIS:")
    print(f"   Délai moyen à l'arrivée: {avg_delay_arr:.2f} minutes")
    print(f"   Délai médian à l'arrivée: {median_delay_arr:.2f} minutes")
    print(f"   Délai moyen au départ: {avg_delay_dep:.2f} minutes")
    
    if 'DISTANCE' in flights.columns:
        print(f"\n✈️ DISTANCE:")
        print(f"   Distance moyenne: {flights['DISTANCE'].mean():.0f} miles")
        print(f"   Distance totale: {flights['DISTANCE'].sum():,.0f} miles")
    
    return {
        'total_vols': total_vols,
        'cancelled': cancelled,
        'delayed': delayed,
        'ontime': ontime,
        'avg_delay_arr': avg_delay_arr
    }
```

```python
# 5. ANALYSES PAR DIMENSION
# ============================================================================

def analyze_airlines(flights):
    """Analyse par compagnie aérienne"""
    print("\n" + "="*80)
    print("ANALYSE PAR COMPAGNIE AÉRIENNE")
    print("="*80)
    
    airline_stats = flights.groupby(['AIRLINE', 'AIRLINE_NAME']).agg({
        'ARRIVAL_DELAY': ['mean', 'median'],
        'CANCELLED': 'sum',
        'AIRLINE': 'count'
    }).round(2)
    
    airline_stats.columns = ['Délai_Moyen', 'Délai_Médian', 'Annulations', 'Nombre_Vols']
    airline_stats = airline_stats.reset_index()
    airline_stats['Taux_Annulation_%'] = (airline_stats['Annulations'] / airline_stats['Nombre_Vols'] * 100).round(2)
    airline_stats = airline_stats.sort_values('Délai_Moyen', ascending=False)
    
    print("\nTop 10 des compagnies avec le plus de retards:")
    print(airline_stats.head(10).to_string())
    
    # Sauvegarde
    airline_stats.to_csv('analyse_compagnies.csv', index=False)
    print("\n📄 Fichier sauvegardé: analyse_compagnies.csv")
    
    # Graphique
    top10 = airline_stats.head(10)
    fig = px.bar(top10, x='AIRLINE', y='Délai_Moyen',
                 hover_data=['AIRLINE_NAME', 'Nombre_Vols'],
                 title='Top 10 des compagnies avec le plus de retards',
                 labels={'Délai_Moyen': 'Délai moyen (min)', 'AIRLINE': 'Code compagnie'},
                 color='Délai_Moyen', color_continuous_scale='Reds')
    fig.write_html('analyse_compagnies.html')
    print("📊 Graphique sauvegardé: analyse_compagnies.html")
    
    return airline_stats

def analyze_airports(flights):
    """Analyse par aéroport"""
    print("\n" + "="*80)
    print("ANALYSE PAR AÉROPORT")
    print("="*80)
    
    airport_stats = flights.groupby(['ORIGIN_AIRPORT', 'ORIGIN_NAME', 'ORIGIN_CITY']).agg({
        'DEPARTURE_DELAY': 'mean',
        'ARRIVAL_DELAY': 'mean',
        'CANCELLED': 'sum',
        'ORIGIN_AIRPORT': 'count'
    }).round(2)
    
    airport_stats.columns = ['Délai_Départ', 'Délai_Arrivée', 'Annulations', 'Nombre_Vols']
    airport_stats = airport_stats.reset_index()
    airport_stats = airport_stats.sort_values('Nombre_Vols', ascending=False)
    
    print("\nTop 10 des aéroports les plus actifs:")
    print(airport_stats.head(10).to_string())
    
    # Sauvegarde
    airport_stats.to_csv('analyse_aeroports.csv', index=False)
    print("\n📄 Fichier sauvegardé: analyse_aeroports.csv")
    
    return airport_stats

def analyze_temporal(flights):
    """Analyse temporelle"""
    print("\n" + "="*80)
    print("ANALYSE TEMPORELLE")
    print("="*80)
    
    # Par mois
    monthly_stats = flights.groupby(['MONTH', 'MONTH_NAME']).agg({
        'ARRIVAL_DELAY': 'mean',
        'CANCELLED': 'sum',
        'MONTH': 'count'
    }).round(2)
    
    monthly_stats.columns = ['Délai_Moyen', 'Annulations', 'Nombre_Vols']
    monthly_stats = monthly_stats.reset_index()
    monthly_stats = monthly_stats.sort_values('MONTH')
    
    print("\nStatistiques mensuelles:")
    print(monthly_stats.to_string())
    
    # Par saison
    season_stats = flights.groupby('SEASON').agg({
        'ARRIVAL_DELAY': 'mean',
        'CANCELLED': ['sum', 'count']
    }).round(2)
    
    season_stats.columns = ['Délai_Moyen', 'Annulations', 'Total_Vols']
    season_stats = season_stats.reset_index()
    season_stats['Taux_Annulation_%'] = (season_stats['Annulations'] / season_stats['Total_Vols'] * 100).round(2)
    
    print("\nStatistiques par saison:")
    print(season_stats.to_string())
    
    # Sauvegarde
    monthly_stats.to_csv('analyse_mensuelle.csv', index=False)
    season_stats.to_csv('analyse_saisonniere.csv', index=False)
    print("\n📄 Fichiers sauvegardés: analyse_mensuelle.csv, analyse_saisonniere.csv")
    
    # Graphique
    fig = make_subplots(rows=2, cols=1,
                        subplot_titles=('Volume de vols par mois', 'Délai moyen par mois'))
    
    fig.add_trace(go.Bar(x=monthly_stats['MONTH_NAME'], y=monthly_stats['Nombre_Vols'],
                         name='Nombre de vols', marker_color='lightblue'), row=1, col=1)
    fig.add_trace(go.Scatter(x=monthly_stats['MONTH_NAME'], y=monthly_stats['Délai_Moyen'],
                            mode='lines+markers', name='Délai moyen', 
                            line=dict(color='red', width=3)), row=2, col=1)
    
    fig.update_layout(height=800, showlegend=False)
    fig.write_html('analyse_temporelle.html')
    print("📊 Graphique sauvegardé: analyse_temporelle.html")
    
    return monthly_stats, season_stats

def analyze_routes(flights):
    """Analyse des routes"""
    print("\n" + "="*80)
    print("ANALYSE DES ROUTES")
    print("="*80)
    
    routes = flights.groupby(['ORIGIN_AIRPORT', 'DESTINATION_AIRPORT',
                              'ORIGIN_NAME', 'DESTINATION_NAME']).size().reset_index(name='Nombre_Vols')
    routes['Route'] = routes['ORIGIN_AIRPORT'] + ' → ' + routes['DESTINATION_AIRPORT']
    routes = routes.sort_values('Nombre_Vols', ascending=False).head(20)
    
    print("\nTop 20 des routes les plus fréquentées:")
    print(routes[['Route', 'ORIGIN_NAME', 'DESTINATION_NAME', 'Nombre_Vols']].to_string())
    
    # Sauvegarde
    routes.to_csv('analyse_routes.csv', index=False)
    print("\n📄 Fichier sauvegardé: analyse_routes.csv")
    
    return routes
```

```python
# 6. CORRÉLATIONS
# ============================================================================

def analyze_correlations(flights):
    """Analyse des corrélations"""
    print("\n" + "="*80)
    print("ANALYSE DES CORRÉLATIONS")
    print("="*80)
    
    numeric_cols = ['DEPARTURE_DELAY', 'ARRIVAL_DELAY', 'DISTANCE', 'AIR_TIME',
                    'ELAPSED_TIME', 'MONTH', 'DAY', 'DAY_OF_WEEK']
    numeric_cols = [col for col in numeric_cols if col in flights.columns]
    
    corr_matrix = flights[numeric_cols].corr().round(2)
    
    print("\nMatrice de corrélation:")
    print(corr_matrix.to_string())
    
    # Graphique
    plt.figure(figsize=(12, 10))
    sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0,
                square=True, linewidths=1, fmt='.2f')
    plt.title('Matrice de corrélation', fontsize=16, fontweight='bold')
    plt.tight_layout()
    plt.savefig('correlation_matrix.png', dpi=300, bbox_inches='tight')
    print("\n📊 Graphique sauvegardé: correlation_matrix.png")
    plt.close()
    
    return corr_matrix
```

```python
# 7. GÉNÉRATION DU RAPPORT
# ============================================================================

def generate_report(stats, airline_stats, monthly_stats, season_stats):
    """Génère un rapport texte complet"""
    print("\n" + "="*80)
    print("GÉNÉRATION DU RAPPORT FINAL")
    print("="*80)
    
    report = []
    report.append("="*80)
    report.append("RAPPORT D'ANALYSE EXPLORATOIRE DES DONNÉES DE VOLS")
    report.append("="*80)
    report.append(f"Date de génération: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
    report.append("\n" + "="*80)
    report.append("RÉSUMÉ EXÉCUTIF")
    report.append("="*80)
    
    report.append(f"\n📊 VOLUME:")
    report.append(f"   • Total de vols analysés: {stats['total_vols']:,}")
    
    report.append(f"\n⏰ PERFORMANCE:")
    report.append(f"   • Taux de ponctualité: {stats['ontime']/stats['total_vols']*100:.1f}%")
    report.append(f"   • Taux d'annulation: {stats['cancelled']/stats['total_vols']*100:.2f}%")
    report.append(f"   • Délai moyen: {stats['avg_delay_arr']:.2f} minutes")
    
    report.append(f"\n🏆 TOP PERFORMERS:")
    if len(airline_stats) > 0:
        best = airline_stats.iloc[-1]
        worst = airline_stats.iloc[0]
        report.append(f"   • Meilleure compagnie: {best['AIRLINE_NAME']} ({best['Délai_Moyen']:.2f} min)")
        report.append(f"   • Compagnie à améliorer: {worst['AIRLINE_NAME']} ({worst['Délai_Moyen']:.2f} min)")
    
    report.append(f"\n📅 TENDANCES SAISONNIÈRES:")
    if len(season_stats) > 0:
        best_season = season_stats.loc[season_stats['Délai_Moyen'].idxmin()]
        worst_season = season_stats.loc[season_stats['Délai_Moyen'].idxmax()]
        report.append(f"   • Meilleure saison: {best_season['SEASON']} ({best_season['Délai_Moyen']:.2f} min)")
        report.append(f"   • Saison la plus problématique: {worst_season['SEASON']} ({worst_season['Délai_Moyen']:.2f} min)")
    
    report.append("\n" + "="*80)
    report.append("FICHIERS GÉNÉRÉS:")
    report.append("="*80)
    report.append("   • analyse_compagnies.csv")
    report.append("   • analyse_aeroports.csv")
    report.append("   • analyse_mensuelle.csv")
    report.append("   • analyse_saisonniere.csv")
    report.append("   • analyse_routes.csv")
    report.append("   • correlation_matrix.png")
    report.append("   • analyse_compagnies.html")
    report.append("   • analyse_temporelle.html")
    report.append("   • rapport_exploration.txt")
    
    report.append("\n" + "="*80)
    report.append("FIN DU RAPPORT")
    report.append("="*80)
    
    # Sauvegarde du rapport
    with open('rapport_exploration.txt', 'w', encoding='utf-8') as f:
        f.write('\n'.join(report))
    
    print("\n📄 Rapport complet sauvegardé: rapport_exploration.txt")
    print('\n'.join(report))
```

```python
# 8. FONCTION PRINCIPALE
# ============================================================================

def main():
    """Fonction principale qui exécute toute l'analyse"""
    
    # Chargement
    flights, airlines, airports = load_data()
    if flights is None:
        return
    
    # Exploration initiale
    explore_structure(flights, "FLIGHTS")
    analyze_missing(flights, "flights")
    
    # Enrichissement
    flights = enrich_data(flights, airlines, airports)
    
    # Analyses
    stats = global_statistics(flights)
    airline_stats = analyze_airlines(flights)
    airport_stats = analyze_airports(flights)
    monthly_stats, season_stats = analyze_temporal(flights)
    routes = analyze_routes(flights)
    corr_matrix = analyze_correlations(flights)
    
    # Rapport final
    generate_report(stats, airline_stats, monthly_stats, season_stats)
    
    print("\n" + "="*80)
    print("✅ ANALYSE TERMINÉE AVEC SUCCÈS")
    print("="*80)
    print(f"Fin de l'analyse: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
    print("="*80 + "\n")
```
